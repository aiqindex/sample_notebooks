# How to look up price Data for finnhub Data

```python
import os
import pandas as pd
from aiq_strategy_robot.data.data_accessor import DAL
```

```python
## load wrapper
from aiq_strategy_robot.data.FINNHUB import (
    load_finnhub_equity_prices,
    load_finnhub_dividends,
    load_finnhub_stock_splits,
    load_finnhub_equity_data
)
```

```python
sdh = DAL()
```

## stock prices

```python
load_finnhub_equity_prices(
    sdh,
    symbols='8035.T', 
    freq='60', 
    start_datetime='2023-12-29', 
    end_datetime='2024-02-01'
).retrieve()
```

```python
load_finnhub_equity_prices(
    sdh,
    symbols='AMZN', 
    freq='D', 
    start_datetime='2023-12-29', 
    end_datetime='2024-02-01'
).retrieve()
```

## dividends

```python
load_finnhub_dividends(
    sdh,
    symbols='8035.T',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

```python
load_finnhub_dividends(
    sdh,
    symbols='AMZN',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

## stock splits

```python
load_finnhub_stock_splits(
    sdh,
    symbols='8035.T',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

```python
load_finnhub_stock_splits(
    sdh,
    symbols='GOOG',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

## Load Market Data, Dividends, and Split ratios in one call

```python
load_finnhub_equity_data(
    sdh,
    symbols=['AMZN', '8035.T'],
    freq='D',
    # split_adjust=False, 👈 Apply stock split adjustment. Default = TRUE (applied)
    # div_adjust=False,   👈 Apply dividend adjustment. Default = TRUE (applied)
    start_datetime='2015-09-30', 
    end_datetime='2024-01-01'
).retrieve()
```

## Appendix 

> Handler.load()

Can also be obtained directly from the handler without using the wrapper.<br>
In this case, `data_source` and `data_type` must be specified.

```python
sdh.load(
    'finnhub', 
    data_type='equity_prices',
    symbols='8035.T', 
    freq='60', 
    start_datetime='2023-12-29', 
    end_datetime='2024-02-01'
).retrieve()
```

```python
sdh.load(
    'finnhub', 
    data_type='dividends',
    symbols='8035.T',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

```python
sdh.load(
    'finnhub', 
    data_type='stock_splits',
    symbols='8035.T',
    start_datetime='2015-09-30', 
    end_datetime='2023-12-31'
).retrieve()
```

```python
sdh.load(
    'finnhub', 
    data_type='equity_data',
    symbols=['AMZN', '8035.T'],
    freq='D',
    start_datetime='2015-09-30', 
    end_datetime='2024-01-01'
).retrieve()
```
